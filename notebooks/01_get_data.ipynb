```python
import pandas as pd
import zipfile
import os

print("Libraries loaded successfully!")
```

```python
# Path to the data folder
DATA_DIR = "../data/"

# All of the  files
files = {
    "patent":     DATA_DIR + "g_patent.tsv.zip",
    "abstract":   DATA_DIR + "g_patent_abstract.tsv.zip",
    "inventor":   DATA_DIR + "g_inventor_disambiguated.tsv.zip",
    "assignee":   DATA_DIR + "g_assignee_disambiguated.tsv.zip",
    "location":   DATA_DIR + "g_location_disambiguated.tsv.zip",
    "pers_inv":   DATA_DIR + "g_persistent_inventor.tsv.zip",
    "pers_ass":   DATA_DIR + "g_persistent_assignee.tsv.zip",
}

# Confirming that all the  files exist
for name, path in files.items():
    exists = os.path.exists(path)
    print(f"{name}: {' Found' if exists else 'NOT FOUND'} in {path}")
```

```python
# Peeking Inside Each File
for name, path in files.items():
    print(f"\n--- {name} ---")
    with zipfile.ZipFile(path) as z:
        inner_file = z.namelist()[0]
        df = pd.read_csv(z.open(inner_file), sep='\t', nrows=3, low_memory=False)
        print(f"Columns: {list(df.columns)}")
        print(f"Shape preview: {df.shape}")
```

```python
# Checking row counts
for name, path in files.items():
    with zipfile.ZipFile(path) as z:
        inner_file = z.namelist()[0]
        df = pd.read_csv(z.open(inner_file), sep='\t', nrows=None, low_memory=False)
        print(f"{name}: {len(df):,} rows")
```

```python
# Previewing Patent File
with zipfile.ZipFile(files["patent"]) as z:
    inner_file = z.namelist()[0]
    df_patent = pd.read_csv(z.open(inner_file), sep='\t', nrows=5, low_memory=False)
    display(df_patent[['patent_id','patent_type','patent_date','patent_title']])
```

```python
#Previewing Inventor File
with zipfile.ZipFile(files["inventor"]) as z:
    inner_file = z.namelist()[0]
    df_inventor = pd.read_csv(z.open(inner_file), sep='\t', nrows=5, low_memory=False)
    display(df_inventor[['patent_id','inventor_id','disambig_inventor_name_first','disambig_inventor_name_last','location_id']])
```

```python
# Previewing Assignee File
with zipfile.ZipFile(files["assignee"]) as z:
    inner_file = z.namelist()[0]
    df_assignee = pd.read_csv(z.open(inner_file), sep='\t', nrows=5, low_memory=False)
    display(df_assignee[['patent_id','assignee_id','disambig_assignee_organization','assignee_type']])
```

```python
# Previewing Location File
with zipfile.ZipFile(files["location"]) as z:
    inner_file = z.namelist()[0]
    df_location = pd.read_csv(z.open(inner_file), sep='\t', nrows=5, low_memory=False)
    display(df_location[['location_id','disambig_city','disambig_state','disambig_country']])
```

```python
#Previewing Abstract File
with zipfile.ZipFile(files["abstract"]) as z:
    inner_file = z.namelist()[0]
    df_abstract = pd.read_csv(z.open(inner_file), sep='\t', nrows=5, low_memory=False)
    display(df_abstract[['patent_id','patent_abstract']])
```
